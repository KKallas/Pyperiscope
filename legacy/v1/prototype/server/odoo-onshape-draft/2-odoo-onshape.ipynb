# 5. Get product dimensions

2. Copy the selected text by ```ctrl```+```c```
3. Vefify text format "000x000x000" and no other characters
4. Deselect the window for normal return by clicking y+100

````python
# comment: Automated step generated with pyPeriscope V2 
payload = ''''''
step = Scope(saved_dict=dill.loads(codecs.decode(payload.encode(), 'base64')))
step.render_preview()
````

````python
step.find()
````

````python
if step.found_locations:
    pyautogui.click(step.get_locations()[0], clicks=3, interval=0.25)
    pyautogui.hotkey('ctrl','c')
    automation_payload["product_dimensions"] = pyperclip.paste().split('x')
    if not(len(automation_payload["product_dimensions"]) == 3):
        raise ValueError("dimensions must be in format 000x000x000")
    try:
        product_dimensions_int = [int(x) for x in automation_payload["product_dimensions"]]
    except ValueError as e:
        raise ValueError("Invalid value found - all elements must be numeric strings")

    print("Found dimensions: "+str(automation_payload["product_dimensions"]))
    # unactivate the text box
    pyautogui.click((step.get_locations()[0][0],step.get_locations()[0][1]-100))
````
